## Making a Basic Chatbot 

```python
import requests
import random
from datetime import datetime


def get_weather(info):
    city = input("Enter city name: ")

    try:
        # City ki location nikalo
        geo = requests.get(
            "https://geocoding-api.open-meteo.com/v1/search",
            params={"name": city},
            timeout=10
        ).json()

        if "results" not in geo:
            return "City not found!"

        place = geo["results"][0]

        # Current weather lao
        response = requests.get(
            "https://api.open-meteo.com/v1/forecast",
            params={
                "latitude": place["latitude"],
                "longitude": place["longitude"],
                "current": "temperature_2m,relative_humidity_2m,wind_speed_10m"
            },
            timeout=10
        )
        response.raise_for_status()
        current = response.json()["current"]

        temp = current["temperature_2m"]
        humidity = current["relative_humidity_2m"]
        wind = current["wind_speed_10m"]
        name = place["name"]

        if info == "humidity":
            return f"Humidity in {name} is {humidity}%"
        elif info == "wind":
            return f"Wind speed in {name} is {wind} km/h"
        elif info == "hot":
            if temp >= 35:
                return f"Yes, it is very hot in {name} ({temp}°C)."
            elif temp <= 15:
                return f"No, it is cold in {name} ({temp}°C)."
            else:
                return f"It is pleasant in {name} ({temp}°C)."
        else:
            return f"Temperature in {name} is {temp}°C"

    except (requests.RequestException, KeyError, ValueError):
        return "Unable to get weather right now!"



def TemperatureBot(message):
    message = message.lower().strip()
    words = message.split()

    # Greetings
    if any(w in words for w in ["hello", "hi", "hey"]):
        return "Hello! How can I help you?"

    elif "good morning" in message:
        return "Good morning! Hope you have a great day."

    elif "how are you" in message:
        return "I am fine! What about you?"

    # Bot info
    elif "your name" in message:
        return "My name is Temperature Bot."

    elif "who made you" in message:
        return "I was created by my developer using Python."

    elif "help" in words or "what can you do" in message:
        return ("I can tell you temperature, humidity, wind speed, "
                "time, date and jokes. Try: 'temperature', 'humidity', "
                "'wind speed', 'is it hot', 'time', 'date'.")

    # Weather
    elif "rain" in message:
        return "Sorry, currently i can only check temperature, humidity and wind."

    elif "humidity" in message:
        return get_weather("humidity")

    elif "wind" in message:
        return get_weather("wind")

    elif "hot" in words or "cold" in words:
        return get_weather("hot")

    elif "weather" in message or "temperature" in message:
        return get_weather("temperature")

    # Time and date
    elif "time" in words:
        return "Current time is " + datetime.now().strftime("%I:%M %p")

    elif "date" in words:
        return "Today's date is " + datetime.now().strftime("%d %B %Y")

    # Ending
    elif "thank" in message:
        return "You're welcome!"

    elif "bye" in words:
        return "Bye! Have a nice day."

    else:
        return "Sorry, I don't understand. Type 'help' to see what I can do."


# Main chatbot loop
while True:
    user = input("User: ")

    if user.lower() == "quit":
        print("Temperature Bot: Goodbye!")
        break

    answer = TemperatureBot(user)
    print("Temperature Bot:", answer)
```
